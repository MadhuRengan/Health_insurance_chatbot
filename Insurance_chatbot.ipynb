```python
import json
import os
from langchain_google_genai import ChatGoogleGenerativeAI #gemini
#rag
from langchain_community.embeddings import HuggingFaceEmbeddings#word vector
from langchain_community.vectorstores import FAISS#chunk table
from langchain_text_splitters import CharacterTextSplitter#chunking
from langchain_classic.chains import RetrievalQA#similarity
```

```python
with open ("/content/knowledge_base (1).json", "r") as f:
  txt = json.load(f)
```

```python
txt
```

```python
#chunking
text_splitter = CharacterTextSplitter(chunk_size=500, chunk_overlap = 50)
chunks = text_splitter.split_text(json.dumps(txt))

#embedding
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vector_store = FAISS.from_texts(chunks,embeddings)
```

```python
prompt = """
You are a helpful and professional customer support bot. Your goal is to assist users with health insurance queries such as coverage, exclusions, claims, premiums, and riders using ONLY the provided Knowledge Base context.

**Guidelines:**
1. **Source-Based Answers:** Use the provided Knowledge Base snippets to answer. If the answer isn't in the context, politely state: "I'm sorry, I don't have information on that. Would you like me to connect you with a human representative?"
2. **Conciseness:** Keep responses under 3 sentences unless explaining a multi-step process (like claim submission).
3. **Tone:** Be empathetic and professional. Use "we" when referring to the insurer (e.g., "We cover cataract surgery under day care procedures...").
4. **Formatting:** Use bullet points for steps, timelines, or lists of inclusions/exclusions to make responses easy to read.
5. **Safety:** Do not invent policies, benefits, or discount codes. Stick strictly to the facts in the KB.
6. **Clarity:** If multiple KB entries are relevant, combine them into a single cohesive answer without repeating verbatim text.
"""
```

```python
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
# Retrieve the API key from Colab secrets (or define it directly, but this is less secure)
GOOGLE_API_KEY = userdata.get('api')

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash",temperature=0.5,google_api_key=GOOGLE_API_KEY)
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=vector_store.as_retriever(),

)
```

```python
query = "please tell about premium policy"

response = qa_chain.invoke(prompt +"_"+ query)
```
